```python
data['tpep_pickup_datetime'] = pd.to_datetime(data['tpep_pickup_datetime'])
data['tpep_dropoff_datetime'] = pd.to_datetime(data['tpep_dropoff_datetime'])
```

```python
from haversine import haversine
def get_distance(row):
    pickup = (row['pickup_latitude'], row['pickup_longitude'])
    dropoff = (row['dropoff_latitude'], row['dropoff_longitude'])
    return haversine(pickup, dropoff)

data['trip_distance_km'] = data.apply(get_distance, axis=1)
```

```python
data.head()
```

```python
data['pickup_hour'] = data['tpep_pickup_datetime'].dt.hour
data['pickup_day'] = data['tpep_pickup_datetime'].dt.day_name()
data['is_weekend'] = data['pickup_day'].isin(['Saturday','Sunday']).astype(int)
data['am_pm'] = data['pickup_hour'].apply(lambda x: 'AM' if x < 12 else 'PM')
data['is_night'] = ((data['pickup_hour'] >= 22) | (data['pickup_hour'] <= 5)).astype(int)
```

```python
data.head()
```

```python
data['trip_duration_min'] = (
    data['tpep_dropoff_datetime'] - data['tpep_pickup_datetime']
).dt.total_seconds() / 60
```

```python
Numerical_cols = ['passenger_count','trip_distance_km', 'trip_duration_min','pickup_hour','total_amount']

plt.figure(figsize=(15,7))
plt.suptitle("Boxplot before outliers removal")
for i in range(0, len(Numerical_cols)):
  plt.subplot(1,8, i+1)
  sns.boxplot(y=data_filtered[Numerical_cols[i]],color= 'purple',)
  plt.tight_layout()
```

```python
import numpy as np

for col in Numerical_cols:
    Q1 = data_filtered[col].quantile(0.25)
    Q3 = data_filtered[col].quantile(0.75)
    IQR = Q3 - Q1

    Lower_bound = Q1 - 1.5 * IQR
    Upper_bound = Q3 + 1.5 * IQR

    data_filtered[col] = np.where(data_filtered[col] < Lower_bound, Lower_bound, data_filtered[col])
    data_filtered[col] = np.where(data_filtered[col] > Upper_bound, Upper_bound, data_filtered[col])
```

# **Model Building**

```python
from xgboost import XGBRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
X = cleaned_data.drop('total_amount',axis=1)
y = cleaned_data['total_amount']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree Regressor": DecisionTreeRegressor(),
    "Random Forest Regressor": RandomForestRegressor(),
    "XGBoost Regressor": XGBRegressor()
}
# Train and evaluate models
results = {}

for name, model in models.items():
    model.fit(X_train, y_train)  # Train the model
    y_pred = model.predict(X_test)  # Predict on test data

    # Compute evaluation metrics
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    results[name] = {
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2 Score": r2
    }

# Convert results to DataFrame
results_df = pd.DataFrame(results).T
results_df
```

**Cross Validation**

```python
from xgboost import XGBRegressor
from sklearn.model_selection import cross_val_score, KFold
import numpy as np

# Define the model
xgb = XGBRegressor(n_estimators=100, random_state=42)

# Define K-Fold Cross-Validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Perform Cross-Validation
cv_scores = cross_val_score(xgb, X, y, cv=kf, scoring='r2')

# Print all 5 scores
print(f"All 5-Fold Scores: {cv_scores}")
print()
# Compare Mean and Standard Deviation
print(f"Mean R2 Score: {np.mean(cv_scores):.4f}")
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import pickle
import numpy as np

# Load model and features
with open("xgboost_fare_model.pkl", "rb") as file:
    model = pickle.load(file)

with open("feature_columns.pkl", "rb") as file:
    feature_columns = pickle.load(file)

st.title("🚕 Taxi Fare Prediction App")

st.write("Enter trip details to predict the total fare amount")

# ---- User Inputs ----
passenger_count = st.number_input("Passenger Count", min_value=1, max_value=6, value=1)
trip_distance_km = st.number_input("Trip Distance (km)", min_value=0.1, value=2.0)
trip_duration_min = st.number_input("Trip Duration (minutes)", min_value=1.0, value=10.0)
pickup_hour = st.slider("Pickup Hour", 0, 23, 12)
is_night = st.selectbox("Is Night Ride?", [0, 1])

pickup_day = st.selectbox("Pickup Day", ["Tuesday", "Thursday"])
am_pm = st.selectbox("AM / PM", ["AM", "PM"])

# ---- Feature Engineering ----
input_data = {
    "passenger_count": passenger_count,
    "trip_distance_km": trip_distance_km,
    "trip_duration_min": trip_duration_min,
    "pickup_hour": pickup_hour,
    "is_night": is_night,
    "pickup_day_Thursday": 1 if pickup_day == "Thursday" else 0,
    "pickup_day_Tuesday": 1 if pickup_day == "Tuesday" else 0,
    "am_pm_AM": 1 if am_pm == "AM" else 0,
    "am_pm_PM": 1 if am_pm == "PM" else 0
}

# Convert to DataFrame
input_df = pd.DataFrame([input_data])

# Ensure correct column order
input_df = input_df.reindex(columns=feature_columns, fill_value=0)

# ---- Prediction ----
if st.button("Predict Fare"):
    prediction = model.predict(input_df)[0]
    st.success(f"💰 Estimated Total Fare: ${prediction:.2f}")
```
